# Generate Metric Cache

Calls `run_metric_caching.py` directly from the notebook.

## 1. Config

```python
import os
from pathlib import Path

NAVSIM_ROOT         = Path("/home/weiyu/workspace/navsim")
OPENSCENE_DATA_ROOT = Path(os.environ.get("OPENSCENE_DATA_ROOT", NAVSIM_ROOT / "dataset"))
NAVSIM_EXP_ROOT     = Path(os.environ.get("NAVSIM_EXP_ROOT",   NAVSIM_ROOT / "exp"))

DATA_SPLIT        = "mini"
METRIC_CACHE_PATH = NAVSIM_EXP_ROOT / "metric_cache"
SCRIPT_PATH       = NAVSIM_ROOT / "navsim/planning/script/run_metric_caching.py"

print(f"Script          : {SCRIPT_PATH}  (exists: {SCRIPT_PATH.exists()})")
print(f"Data split      : {DATA_SPLIT}")
print(f"Metric cache out: {METRIC_CACHE_PATH}")
```

## 2. Run Metric Caching Script

```python
import subprocess, sys

cmd = [
    sys.executable,
    str(SCRIPT_PATH),
    f"train_test_split={DATA_SPLIT}",
    f"metric_cache_path={METRIC_CACHE_PATH}",
    "worker=sequential",
]

print("Running:", " ".join(cmd))
print("-" * 60)

env = os.environ.copy()
env["OPENSCENE_DATA_ROOT"] = str(OPENSCENE_DATA_ROOT)
env["NAVSIM_EXP_ROOT"]     = str(NAVSIM_EXP_ROOT)

result = subprocess.run(cmd, env=env, cwd=str(NAVSIM_ROOT))

print("-" * 60)
print(f"Return code: {result.returncode}")
```
